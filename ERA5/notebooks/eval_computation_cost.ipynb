# Visualize Model Predictions

This notebook loads a trained model and visualizes its predictions on a new sample.

```python
import sys
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
df_pre = pd.read_csv("../scripts/computation_times_preproc.csv", index_col=False)
df_emb = pd.read_csv("../scripts/computation_times_embed.csv", index_col=False)

df = pd.DataFrame(df_pre)
df['reduction_time'] = df_emb['reduction_time']
df
```

```python
k_list = []
N_list = []
preproc_time = []
preproc_std = []
forward_time = []
forward_std = []
preproc_string = []
forward_string = []

for N in df.N.unique():
    for k in df.k.unique():
        N_list.append(N)        
        k_list.append(k)
    
        preproc_values = df.loc[(df.k == k) & (df.N == N)].preprocessing_time
        forward_values = df.loc[(df.k == k) & (df.N == N)].reduction_time
        
        preproc_time.append(preproc_values.mean())
        preproc_std.append(preproc_values.std())
        preproc_string.append(f'{preproc_values.mean()*1000:.2f} &$\\pm$ {preproc_values.std()*1000:.2f}')
        
        forward_time.append(forward_values.mean())
        forward_std.append(forward_values.std())
        forward_string.append(f'{forward_values.mean()*1000:.2f} $\\pm$ {forward_values.std()*1000:.2f}')

df_processed = pd.DataFrame({
    'N': N_list,
    'k': k_list,
    'Preprocessing [ms]': preproc_string,
    'Forward Pass [ms]': forward_string,
})

print(df_processed.to_latex(index=False))
```

```python
fig, ax = plt.subplots()


for k in df.k.unique():
    list = []
    for N in df.N.unique():
        value = df.loc[(df.k == k) & (df.N == N)].reduction_time.mean()
        list.append(value)

    ax.plot(df.N.unique(), list, label=f'k = {k}')
    ax.set_yscale("log")
    ax.legend()


fig.show()
```
